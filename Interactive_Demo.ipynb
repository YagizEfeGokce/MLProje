# Interactive University Ranking Predictor

This notebook allows you to interactively predict a university's global ranking category using the pre-trained XGBoost model.

**Instructions:**
1. Run the cell below.
2. Enter the ranking values for each metric when prompted.
3. View the predicted category.

```python
import joblib
import numpy as np
import os

def interactive_predict():
    if not os.path.exists("results/xgboost_model.pkl"):
        print("Error: Model not found. Please run 'src/main.py' first.")
        return

    print("Loading Model...")
    model = joblib.load("results/xgboost_model.pkl")
    scaler = joblib.load("results/scaler.pkl")
    
    print("Model Loaded Successfully!\n")
    print("--- Enter University Metrics ---")
    
    try:
        q_ed = float(input("Quality of Education Rank (e.g. 355): "))
        alumni = float(input("Alumni Employment Rank (e.g. 478): "))
        q_fac = float(input("Quality of Faculty Rank (e.g. 218): "))
        pubs = float(input("Publications Rank (e.g. 995): "))
        inf = float(input("Influence Rank (e.g. 850): "))
        cite = float(input("Citations Rank (e.g. 812): "))
        broad = float(input("Broad Impact Rank (e.g. 1000): "))
        patents = float(input("Patents Rank (e.g. 750): "))
        country = float(input("Country Code (Enter 0 if unknown): "))
        
        features = np.array([q_ed, alumni, q_fac, pubs, inf, cite, broad, patents, country]).reshape(1, -1)
        features_scaled = scaler.transform(features)
        
        pred = model.predict(features_scaled)[0]
        
        labels = {0: "Elite (Top 100)", 1: "High (101-500)", 2: "Average (>500)"}
        print(f"\n>>> PREDICTION: {labels[pred]}")
        
    except ValueError:
        print("Invalid input. Please enter numbers only.")

interactive_predict()
```
